```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
def first_func(a, b):
    return a * b
```

```python
period = 11
```

```python
x = []

for i in range(3, period + 1):
    x.append(i)
    
y = []
for arg in x:
    if arg >= 3 and arg <= 11:
        y.append(first_func(3, arg))
        y.reverse()
    
for i in range(5, len(y)):
    if y[i] > 10:
        y[i] = y[i] + 3
        

print(x)
print(y)
```

```python
plt.xlabel("x")
plt.ylabel("f(x)")

y_list = np.arange(0, 20.01, 1)

x_list = np.arange(0, 20.01, 1)

plt.plot(x, y, color = "red", linewidth = 1)

plt.yticks(y_list)
plt.xticks(x_list)

plt.grid()
plt.show()
```
