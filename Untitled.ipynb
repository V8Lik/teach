```python
import matplotlib.pyplot as plt 
import numpy as np
import math
```

```python
def first_func(a, b, c):
    return math.fabs(a * b + c)
```

```python
x = np.arange(3, 11.1 , 1)
y = []

for arg in x:
    y.append(first_func(-3, arg, 21))

print(x)
print(y)
```

```python
plt.xlabel("x")
plt.ylabel("f(x)")

y_list = np.arange(0, 20.01, 1)

x_list = np.arange(0, 20.01, 1)

plt.plot(x, y, color = "red", linewidth = 1)

plt.yticks(y_list)
plt.xticks(x_list)

plt.grid()
plt.show()
```
